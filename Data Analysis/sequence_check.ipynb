```python
#Loop through each row, using a dict to track each ids progression
levelDict = {}
#puzzleDF = puzzleDF.reset_index()
for i, row in puzzleDF.iterrows():
  #If its new, insert into dict as 1
  if row['id'] not in levelDict:
    levelDict[row['id']] = 1

  #Otherwise, check to make sure that it's good
  else:
    if levelDict[row['id']] + 1 != row['level']:
      print("weird value for id " + str(row['id']) + "; expected " + str(levelDict[row['id']] + 1) + " but saw " + str(row['level']))
    levelDict[row['id']] = row['level']
```

Detects what ids were registered multiple times due to duplicate logins.

```python
#Make all of the dataframes from the files
from google.colab import drive
import pandas as pd
drive.mount('/content/drive')

idDF = pd.read_csv('drive/My Drive/Masters Stuff/Data Analysis/Raw CSVs/V2/ids.csv')
```

```python
#Loop through each row, using a dict to track each ids count
countDict = {}
for i, row in idDF.iterrows():
  #If its new, insert into dict as 1
  if row['id'] not in countDict:
    countDict[row['id']] = 1

  #Otherwise, check to make sure that it's good
  else:
    countDict[row['id']] += 1

for i in countDict.keys():
  if countDict[i] > 1:
    print(str(i) + ' has ' + str(countDict[i]) + ' entries')
```
