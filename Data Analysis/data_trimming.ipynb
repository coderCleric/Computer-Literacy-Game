Version keys for each csv version:
- V0: Raw data, doesn't even have bad ids trimmed
- V1: Culled data where invalid ids are removed (based on what ids had permission slips)
- V2: Culled puzzle data that didn't make sense (like students replaying the game)

```python
#Make all of the dataframes from the files
from google.colab import drive
import pandas as pd
drive.mount('/content/drive')

idDF = pd.read_csv('drive/My Drive/Masters Stuff/Data Analysis/Raw CSVs/V0/ids.csv')
puzzleDF = pd.read_csv('drive/My Drive/Masters Stuff/Data Analysis/Raw CSVs/V0/puzzledata.csv')
completionDF = pd.read_csv('drive/My Drive/Masters Stuff/Data Analysis/Raw CSVs/V0/completions.csv')
```

```python
#All of the valid ids
ids = [12111, 12316, 13477, 13823, 15273, 24457, 24648, 28651, 36577, 39316, 41459, 42567, 43268, 43933, 48196, 52978, 54253, 55789, 55988, 58376, 61152, 64497, 65458, 66678,
       68497, 72378, 81555, 82589, 85632, 91459, 94258]
```

```python
#Figure out what invalid ids have been registered
badIds = []
for id in idDF.id:
  if id not in ids:
    badIds.append(id)

print(badIds)
```

```python
#Remove entries with bad ids
for id in badIds:
  idDF = idDF[idDF.id != id]
  puzzleDF = puzzleDF[puzzleDF.id != id]
  completionDF = completionDF[completionDF.id != id]
```

```python
#Actually write the new files
idDF.to_csv('drive/My Drive/Masters Stuff/Data Analysis/Raw CSVs/V1/ids.csv')
puzzleDF.to_csv('drive/My Drive/Masters Stuff/Data Analysis/Raw CSVs/V1/puzzledata.csv')
completionDF.to_csv('drive/My Drive/Masters Stuff/Data Analysis/Raw CSVs/V1/completions.csv')
```
